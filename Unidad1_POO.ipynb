## 1. Del Script Lineal al Diseño Orientado a Objetos

Imaginá que tenés que gestionar usuarios y sus suscripciones en tu startup. Con programación lineal o estructurada simple, usarías diccionarios y funciones sueltas:

```python
usuario = {"nombre": "Sofía", "email": "sofia@email.com", "plan": "Premium"}

    if usuario["plan"] == "Premium":
        return True
    return False
```

Este enfoque se vuelve difícil de mantener a medida que el negocio crece (más tipos de usuarios, métodos de pago, carritos, cupones). Con POO, unificamos los **datos** (atributos) y el **comportamiento** (métodos) en una única entidad: la **Clase**.


### Definición de Clases e Instanciación en Python

Una **Clase** es un molde o plantilla. Un **Objeto** (o instancia) es el elemento concreto creado a partir de ese molde.


````python
# Definición de la clase básica Usuario
class Usuario:
    # El constructor define los atributos iniciales
    def __init__(self, nombre, email):
        self.nombre = nombre       # Atributo público
        self.email = email         # Atributo público
        self.activo = True         # Atributo por defecto

    # Un método define un comportamiento del objeto
    def desactivar(self):
        self.activo = False
        print(f"El usuario {self.nombre} ha sido desactivado.")

# Instanciación de objetos
usuario_1 = Usuario("Sofía G.", "sofia@umsa.edu.ar")
usuario_2 = Usuario("Tomás R.", "tomas@umsa.edu.ar")

# Acceso a atributos y métodos
print(f"Usuario 1: {usuario_1.nombre} ({usuario_1.email}) - Activo: {usuario_1.activo}")
usuario_2.desactivar()
print(f"Usuario 2: {usuario_2.nombre} - Activo: {usuario_2.activo}")
````

4. **Polimorfismo**: Permitir que clases distintas respondan al mismo mensaje/método de formas diferentes.

### 2.1 Abstracción y Encapsulamiento con `@property`

El encapsulamiento evita que se modifiquen atributos de forma inválida. En Python, usamos el prefijo de guión bajo doble o simple (ej. `__saldo`) para indicar atributos privados y exponemos validaciones usando `@property`.


````python
class Transaccion:
    def __init__(self, monto_inicial):
        # Atributo privado/protegido
        if monto_inicial <= 0:
            raise ValueError("El monto inicial debe ser mayor a cero.")
        self.__monto = monto_inicial

    # Getter: Permite leer el atributo protegido
    @property
    def monto(self):
        return self.__monto

    # Setter: Permite modificar el atributo con validación de negocio
    @monto.setter
    def monto(self, nuevo_monto):
        if nuevo_monto <= 0:
            print("Error: El monto de transacción debe ser mayor a cero.")
        else:
            self.__monto = nuevo_monto

# Probando el encapsulamiento
try:
    trx = Transaccion(150.0)
    print(f"Monto de la transacción: ${trx.monto:.2f}")

    # Intentamos poner un monto inválido
    trx.monto = -50.0  # Muestra mensaje de error sin modificar el valor
    print(f"Monto después de intento inválido: ${trx.monto:.2f}")

    # Modificación válida
    trx.monto = 200.0
    print(f"Monto después de actualización válida: ${trx.monto:.2f}")

except ValueError as e:
    print(f"Error en constructor: {e}")
````

# Explicación del Código: Clase `Transaccion` (Encapsulamiento y Propiedades)

Este bloque aborda el pilar de **encapsulamiento** en la Programación Orientada a Objetos mediante el uso de atributos privados, validaciones de reglas de negocio y los decoradores `@property` y `@setter` en Python.

---

### 1. Atributos Privados y *Name Mangling*
* **Doble guion bajo (`self.__monto`):** En Python, anteponer dos guiones bajos activa el mecanismo de *name mangling* (ocultamiento de nombres), transformando internamente la variable para evitar que sea leída o modificada directamente desde el exterior mediante `objeto.__monto`.
* **Propósito:** Protege la integridad del estado interno del objeto, obligando a que cualquier interacción pase por filtros de control.

---

### 2. Validación Temprana en el Constructor (`__init__`)
* **Control de entrada:** Antes de asignar la variable interna, se evalúa si el valor inicial cumple la regla de negocio (debe ser estrictamente superior a cero).
* **Lanzamiento de excepciones (`raise ValueError`):** Si el valor es inválido, se interrumpe la instanciación de forma segura para evitar la existencia de objetos en un estado corrupto o inconsistente.

---

### 3. Lectura Controlada: El Getter (`@property`)
* **Decorador `@property`:** Transforma un método de lectura en una propiedad accesible con sintaxis de atributo tradicional (`trx.monto` en lugar de `trx.get_monto()`).
* **Retorno seguro:** Expone el valor del atributo privado `__monto` sin permitir su sobreescritura directa.

---

### 4. Escritura y Validación: El Setter (`@monto.setter`)
* **Decorador `@monto.setter`:** Intercepta cualquier intento de asignación directa (`trx.monto = nuevo_valor`).
* **Regla de negocio:** Evalúa el nuevo valor recibido:
  * Si no supera el umbral requerido (menor o igual a cero), rechaza la operación, emite una advertencia y preserva el valor anterior.
  * Si cumple la condición, actualiza `self.__monto` con éxito.

---

### 5. Demostración y Manejo de Excepciones (`try / except`)
* **Prueba de instanciación:** Se crea el objeto con un saldo inicial válido y se formatea la salida numérica a dos decimales (`:.2f`).
* **Intento de mutación inválida:** Al intentar asignar un valor negativo, entra en acción el setter bloqueando el cambio y dejando intacto el estado previo.
* **Mutación válida:** Se actualiza el monto satisfactoriamente al cumplir con las restricciones.
* **Captura de error (`except ValueError`):** Bloque de contingencia que previene caídas imprevistas del programa si se intentara instanciar un objeto con un monto inicial no permitido.

### 2.1.1 Getters y Setters Tradicionales vs. El Enfoque Idiomático con `@property`

En lenguajes orientados a objetos como Java, C++ o C#, el estándar para proteger y encapsular variables privadas es escribir métodos explícitos del tipo `get_atributo()` y `set_atributo(nuevo_valor)`.

Aunque en Python se pueden definir métodos tradicionales de este estilo, **el enfoque "pythonico" preferido es el uso de la propiedad (`@property`)**. Este decorador nos permite mantener una sintaxis limpia y directa al acceder o modificar atributos (`objeto.monto = 200`), ejecutando validaciones de negocio por detrás sin romper la interfaz de nuestro código.

A continuación, comparamos ambos enfoques sobre el mismo modelo de negocio (`Transaccion`).

````python
# =====================================================================
# ENFOQUE 1: Getters y Setters Tradicionales (Estilo Java / C++)
# =====================================================================
class TransaccionClasica:
    def __init__(self, monto_inicial: float):
        self.set_monto(monto_inicial)  # Usamos el setter para validar al instanciar

    # Getter tradicional: Método explícito para LEER el atributo
    def get_monto(self) -> float:
        return self._monto

    # Setter tradicional: Método explícito para MODIFICAR y VALIDAR el atributo
    def set_monto(self, nuevo_monto: float) -> None:
        if nuevo_monto <= 0:
            raise ValueError("[ERROR] El monto debe ser estrictamente mayor a 0.")
        self._monto = nuevo_monto


# =====================================================================
# ENFOQUE 2: Enfoque Idiomático de Python (@property)
# =====================================================================
class TransaccionPythonica:
    def __init__(self, monto_inicial: float):
        self.monto = monto_inicial  # Llama internamente a @monto.setter

    # Getter mediante @property
    @property
    def monto(self) -> float:
        return self._monto

    # Setter mediante @monto.setter
    @monto.setter
    def monto(self, nuevo_monto: float) -> None:
        if nuevo_monto <= 0:
            raise ValueError("[ERROR] El monto debe ser strictly mayor a 0.")
        self._monto = nuevo_monto


# =====================================================================
# PRUEBA COMPARATIVA
# =====================================================================
if __name__ == "__main__":
    print("--- 1. Uso de Getters / Setters Tradicionales ---")
    trx_1 = TransaccionClasica(1500.0)
    print("Monto actual (vía get_monto()):", trx_1.get_monto())

    trx_1.set_monto(2000.0)
    print("Monto actualizado (vía set_monto()):", trx_1.get_monto())

    try:
        trx_1.set_monto(-500.0)
    except ValueError as e:
        print(e)


    print("\n--- 2. Uso de @property Idiomático de Python ---")
    trx_2 = TransaccionPythonica(1500.0)
    print("Monto actual (sintaxis natural):", trx_2.monto)  # Llama a @property de forma transparente

    trx_2.monto = 2000.0                                 # Llama a @monto.setter de forma transparente
    print("Monto actualizado (sintaxis natural):", trx_2.monto)

    try:
        trx_2.monto = -500.0
    except ValueError as e:
        print(e)
````

# Explicación del Código: Comparativa de Encapsulamiento en Python

Este bloque compara dos paradigmas para gestionar el acceso y la mutación de atributos protegidos: el patrón clásico de **métodos explícitos (estilo Java/C++)** frente al patrón idiomático o **pythónico mediante `@property`**.

---

### 1. Convención de Atributo Protegido (`_monto`)
* **Guion bajo simple (`_`):** Convención estándar en Python que indica que una variable es de uso interno o protegida. No bloquea el acceso a nivel de intérprete, pero señala formalmente que no debe manipularse directamente desde el exterior.

---

### 2. Enfoque 1: Getters y Setters Tradicionales (`TransaccionClasica`)
* **Métodos explícitos:** Se definen dos funciones separadas, `get_monto()` para la lectura y `set_monto()` para la asignación y validación.
* **Reutilización en el constructor:** El método `__init__` delega la asignación inicial a `self.set_monto(monto_inicial)`, garantizando que las reglas de negocio se apliquen tanto en la creación como en las mutaciones posteriores.
* **Sintaxis de uso:** Obliga al consumidor de la clase a invocar métodos con paréntesis (`trx.get_monto()` y `trx.set_monto(...)`), un patrón rígido heredado de lenguajes estáticamente tipados.

---

### 3. Enfoque 2: Estilo Idiomático de Python (`TransaccionPythonica`)
* **Decorador `@property`:** Convierte el método de lectura en un atributo virtual, permitiendo acceder al valor con sintaxis natural de lectura (`trx.monto`).
* **Decorador `@monto.setter`:** Intercepta la asignación estándar (`trx.monto = valor`), ejecutando la validación de negocio de forma transparente tras bambalinas.
* **Constructor limpio:** Al ejecutar `self.monto = monto_inicial` dentro de `__init__`, se dispara automáticamente el setter decorado, validando la entrada desde el instante de instanciación.

---

### 4. Cuadro Comparativo de Enfoques

| Característica | Enfoque Clásico (`TransaccionClasica`) | Enfoque Pythónico (`TransaccionPythonica`) |
| :--- | :--- | :--- |
| **Sintaxis de lectura** | `trx.get_monto()` (llamada a función) | `trx.monto` (notación de atributo directa) |
| **Sintaxis de escritura** | `trx.set_monto(valor)` | `trx.monto = valor` |
| **Compatibilidad con APIs** | Requiere refactorizar el código cliente si se pasa de atributo público a método | Permite añadir validación posterior sin alterar la interfaz de uso de la clase |
| **Estilo y legibilidad** | Verboso, propio de Java/C++ | Limpio, idiomático y estándar en la comunidad Python |

---

### 5. Demostración y Manejo de Errores (`try / except`)
* **Consistencia en validación:** Ambos enfoques aplican la misma restricción de negocio (`monto > 0`) y disparan una excepción de tipo `ValueError` ante valores no permitidos.
* **Captura de fallos:** Los bloques `try / except` demuestran que, ante intentos de mutación inválidos (valores negativos), la ejecución captura el error de forma controlada sin interrumpir el flujo del programa.

| Criterio | Getters / Setters Tradicionales | Enfoque Pythonico con `@property` |

### 2.2 Herencia y Polimorfismo en Negocios Digitales

Imaginemos diferentes tiers de suscripción en una plataforma SaaS. Queremos calcular el precio final según el descuento de cada plan.


````python
# Clase base (Padre)
class Suscripcion:
    def __init__(self, precio_base):
        self.precio_base = precio_base

    # Método que será sobrescrito (Polimorfismo)
    def calcular_costo(self):
        return self.precio_base

# Clase derivada (Hijo)
class SuscripcionFamiliar(Suscripcion):
    def __init__(self, precio_base, miembros):
        super().__init__(precio_base)
        self.miembros = miembros

    # Sobrescribimos el método con lógica propia
    def calcular_costo(self):
        # Descuento del 10% si son más de 3 miembros
        if self.miembros > 3:
            return (self.precio_base * self.miembros) * 0.9
        return self.precio_base * self.miembros

# Clase derivada (Hijo)
class SuscripcionAnual(Suscripcion):
    def calcular_costo(self):
        # Descuento del 20% por pago anual adelantado (12 meses)
        return (self.precio_base * 12) * 0.8

# Demostración de Polimorfismo
suscripciones = [
    Suscripcion(10.0),                  # Plan Individual mensual: $10.00
    SuscripcionFamiliar(10.0, 4),       # Plan Familiar para 4 con 10% off: $36.00
    SuscripcionAnual(10.0)              # Plan Anual de 12 meses con 20% off: $96.00
]

for idx, sub in enumerate(suscripciones, 1):
    print(f"Costo de suscripción {idx} ({type(sub).__name__}): ${sub.calcular_costo():.2f}")
````

# Explicación del Código: Herencia y Polimorfismo en Python

Este bloque ilustra dos pilares avanzados de la **Programación Orientada a Objetos (POO)**: la **herencia** de clases para reutilizar estructura y lógica base, y el **polimorfismo** mediante la sobrescritura de métodos (*method overriding*).

---

### 1. Clase Base: `Suscripcion`
* **Definición de la jerarquía:** Actúa como la clase padre que establece los atributos y comportamientos comunes a todos los tipos de suscripción.
* **Constructor (`__init__`):** Define el atributo general `self.precio_base`.
* **Método base (`calcular_costo`):** Implementa el cálculo por defecto, devolviendo directamente el precio base mensual sin aplicar multiplicadores ni descuentos.

---

### 2. Clases Derivadas: Herencia y Extensión
* **Herencia directa (`class Hijo(Padre)`):** Tanto `SuscripcionFamiliar` como `SuscripcionAnual` heredan de `Suscripcion`, obteniendo automáticamente sus atributos y capacidades.
* **Uso de `super().__init__(...)`:** En `SuscripcionFamiliar`, la función `super()` invoca el constructor de la clase base para inicializar `precio_base`, evitando código duplicado y agregando a la vez un nuevo atributo específico (`self.miembros`).
* **Herencia sin constructor explícito:** En `SuscripcionAnual`, al no definir un nuevo `__init__`, se reutiliza automáticamente el constructor de la clase base sin requerir código adicional.

---

### 3. Sobrescritura de Métodos (*Method Overriding*)
* **Redefinición de comportamiento:** Ambas clases hijas redefinen el método `calcular_costo()`, reemplazando la implementación original con reglas de negocio particulares:

## 3. Métodos Especiales (Dunder Methods)

En Python, los métodos que empiezan y terminan con doble guión bajo (`__`) se llaman "dunder methods". Nos permiten personalizar el comportamiento básico de los objetos, como su representación en texto.


````python
class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio

    # __str__: Representación amigable para el usuario final (ej. prints)
    def __str__(self):
        return f"{self.nombre} (${self.precio:.2f})"

    # __repr__: Representación técnica para depuración / consola
    def __repr__(self):
        return f"Producto(nombre='{self.nombre}', precio={self.precio})"

# Pruebas
prod = Producto("Laptop Pro 16", 1500.0)
print(str(prod))   # Llama a __str__
print(repr(prod))  # Llama a __repr__
````

### 4.2 Patrón Factory
Define una interfaz para crear un objeto, pero deja que una clase fábrica decida qué clase instanciar en función de algún parámetro (ideal para procesar distintos métodos de pago en un checkout).


````python
class ProcesadorPago:
    def procesar(self, monto):
        pass

class ProcesadorMercadoPago(ProcesadorPago):
    def procesar(self, monto):
        return f"Procesando ${monto:.2f} vía Mercado Pago (QR/Dinero en cuenta)."

class ProcesadorStripe(ProcesadorPago):
    def procesar(self, monto):
        return f"Procesando ${monto:.2f} vía Stripe (Tarjeta de Crédito Internacional)."

# Fábrica de Procesadores
class PagoFactory:
    @staticmethod
    def obtener_procesador(metodo):
        if metodo.lower() == "mercadopago":
            return ProcesadorMercadoPago()
        elif metodo.lower() == "stripe":
            return ProcesadorStripe()
        else:
            raise ValueError(f"Método de pago '{metodo}' no soportado.")

# Uso del Factory
metodo_cliente = "mercadopago"  # Puede venir dinámicamente de una API o Formulario
procesador = PagoFactory.obtener_procesador(metodo_cliente)
print(procesador.procesar(2500.0))
````

Este bloque ilustra la implementación del patrón de diseño creacional **Singleton** mediante la sobrescritura del método especial `__new__`, asegurando que solo exista una única instancia de configuración compartida en toda la aplicación.

---

### 1. Atributo de Clase (`_instancia = None`)
* **Estado global compartido:** Se declara a nivel de clase para almacenar la referencia al objeto único una vez que se cree.
* **Convención protegida (`_`):** El guion bajo simple indica que es una variable de control interno, evitando su manipulación directa desde el código consumidor.

---

### 2. El Método `__new__` vs. `__init__`
* **Control en la creación:** En Python, `__new__` es el método constructor real que asigna memoria y devuelve la instancia del objeto, mientras que `__init__` se limita a inicializar sus valores.
* **El parámetro `cls`:** Hace referencia a la propia clase (`ConfiguracionSaaS`) y no a una instancia particular.
* **Lógica condicional:** Evalúa si `cls._instancia` aún no existe (`is None`):

  * **Llamadas subsiguientes:** Omite la creación y devuelve directamente la instancia preexistente guardada en `cls._instancia`.

---

### 3. Cuadro de Componentes del Patrón

| Elemento | Tipo | Rol en el Patrón Singleton |
| :--- | :--- | :--- |
| **`_instancia`** | Atributo de clase | Mantiene la referencia persistente al único objeto creado en memoria |
| **`__new__(cls)`** | Método de creación | Intercepta la instanciación para restringir la creación a un solo objeto |
| **`super().__new__(cls)`** | Llamada a la clase base | Crea la instancia física en memoria solo la primera vez |
| **`cls._instancia.timeout`** | Atributo de instancia | Almacena parámetros globales accesibles desde cualquier punto del sistema |

---

````python
# --- Resolución Completa del Trabajo Práctico 1 ---

class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio  # Hace uso del setter para validar el valor inicial

    @property
    def precio(self):
        return self.__precio

    @precio.setter
    def precio(self, valor):
        if valor <= 0:
            raise ValueError("El precio debe ser estrictamente mayor a 0.")
        self.__precio = valor

    def __str__(self):
        return f"{self.nombre} (${self.precio:.2f})"


class Carrito:
    def __init__(self):
        self.productos = []
        self.descuento_pct = 0.0

    def agregar_producto(self, producto: Producto):
        self.productos.append(producto)
        print(f"Agregado al carrito: {producto.nombre}")

    def remover_producto(self, producto: Producto):
        if producto in self.productos:
            self.productos.remove(producto)
            print(f"Removido del carrito: {producto.nombre}")
        else:
            print(f"El producto {producto.nombre} no está en el carrito.")

    def aplicar_descuento(self, descuento_pct):
        if 0 <= descuento_pct <= 100:
            self.descuento_pct = descuento_pct
            print(f"Descuento de {descuento_pct}% aplicado correctamente.")
        else:
            raise ValueError("El porcentaje de descuento debe estar entre 0 y 100.")

    @property
    def total(self):
        subtotal = sum(prod.precio for prod in self.productos)
        return subtotal * (1 - (self.descuento_pct / 100))


class Transaccion:
    def __init__(self, monto):
        self.monto = monto

    def calcular_total_final(self):
        return self.monto


class TransaccionTarjeta(Transaccion):
    def calcular_total_final(self):
        # 5% de recargo por comisión de tarjeta
        return self.monto * 1.05


class TransaccionEfectivo(Transaccion):
    def calcular_total_final(self):
        # 10% de descuento por pago en efectivo
        return self.monto * 0.90


class TransaccionFactory:
    @staticmethod
    def crear_transaccion(canal, monto):
        canal_normalizado = canal.strip().lower()
        if canal_normalizado == "tarjeta":
            return TransaccionTarjeta(monto)
        elif canal_normalizado == "efectivo":
            return TransaccionEfectivo(monto)
        else:
            raise ValueError(f"Canal de pago '{canal}' no soportado.")


# --- Flujo de Prueba Completo ---
if __name__ == "__main__":
    print("=== INICIO DE FLUJO DE PRUEBA ===")

    # 1. Creación de productos
    p1 = Producto("Curso de Python Orientado a Objetos", 15000.0)
    p2 = Producto("Ebook de Arquitectura SaaS", 5000.0)

    # 2. Inicialización de Carrito y carga de productos
    carrito = Carrito()
    carrito.agregar_producto(p1)
    carrito.agregar_producto(p2)

    # 3. Aplicar descuento promocional del 10%
    carrito.aplicar_descuento(10)
    monto_carrito = carrito.total
    print(f"Subtotal con descuento del carrito: ${monto_carrito:.2f}")

    # 4. Procesar el pago a través de la fábrica (Tarjeta)
    canal_pago = "Tarjeta"
    transaccion = TransaccionFactory.crear_transaccion(canal_pago, monto_carrito)
    total_final = transaccion.calcular_total_final()
    print(f"Total a pagar ({canal_pago}): ${total_final:.2f}")

    print("=== FIN DEL FLUJO DE PRUEBA ===")
````
