## 🧩 EJERCICIO 1: Programación Orientada a Objetos y Manejo de Errores (35 Puntos)

### Consigna:
El objetivo es modelar la entidad central de una tienda digital con control de inventario básico:

1. **Excepción personalizada:** Cree una excepción llamada `StockInsuficienteError` que herede de `Exception`.
2. **Clase Producto:** Defina la clase `Producto` cuyo método `__init__` reciba y asigne los atributos:
   * `nombre` (str)
   * `precio` (float)
   * `stock` (int)
3. **Representación textual (`__str__`):** Implemente el método especial para que al imprimir el objeto devuelva exactamente el formato:  
   `"Producto: <nombre> | Precio: $<precio con 2 decimales> | Stock: <stock>"`  
   *(Ejemplo: `"Producto: Teclado Mecánico | Precio: $100.00 | Stock: 10"`)*.
4. **Método vender:** Defina el método `vender(cantidad: int) -> float`:
   * Si `cantidad` es mayor al `stock` disponible, debe lanzar `StockInsuficienteError`.
   * Si hay stock suficiente, debe descontar `cantidad` del stock y retornar el costo total (`cantidad * precio`).
