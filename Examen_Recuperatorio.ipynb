### 🧩 Consigna Ejercicio 1:
Se requiere modelar el saldo de una billetera virtual:

1. **Excepción personalizada:** Cree la excepción `FondosInsuficientesError` que herede de `Exception`.
2. **Clase BilleteraDigital:** Defina la clase cuyo método `__init__` reciba:
   * `titular` (str)
   * `saldo_inicial` (float, por defecto `0.0`)
3. **Representación textual (`__str__`):** Debe retornar exactamente:  
   `"Billetera de <titular> | Saldo: $<saldo con 2 decimales>"`  
   *(Ejemplo: `"Billetera de Ana Perez | Saldo: $1500.00"`)*.
4. **Métodos:**
   * `depositar(monto: float) -> float`: Si el monto es mayor a 0, lo suma al saldo y retorna el saldo actualizado. Si es menor o igual a 0, lanza `ValueError("El monto a depositar debe ser positivo")`.
   * `pagar(monto: float) -> float`: Si el monto a pagar supera el saldo disponible, lanza `FondosInsuficientesError("Saldo insuficiente para realizar el pago")`. De lo contrario, descuenta el monto y retorna el saldo restante.

### 🌐 Consigna Ejercicio 2:
Dado un bloque HTML que representa la oferta académica de cursos:

1. Cree la función `filtrar_cursos_intensivos(html: str, min_horas: int = 20) -> list`.
2. Con `BeautifulSoup` inspeccione todos los bloques `<article class="curso">`.
3. Por cada curso extraiga:
   * **Título:** El texto de la etiqueta `<h2 class="titulo">`.
   * **Horas:** El número entero (`int`) dentro de `<span class="duracion">` (ejemplo: si el texto dice `"45 horas"`, debe extraer el número entero `45`).
4. Retorne una lista de diccionarios `[{"titulo": ..., "horas": ...}]` que contenga **únicamente los cursos cuya duración sea mayor o igual a `min_horas`**.

### ⚙️ Consigna Ejercicio 3:
Para el módulo de autenticación de una aplicación:

1. **Modelo UsuarioRegistro:** Defina un modelo heredando de `pydantic.BaseModel` con:
   * `username`: cadena de texto (`str`) con longitud mínima de 4 caracteres (`min_length=4`).
   * `email`: cadena de texto (`str`).
   * `edad`: número entero mayor o igual a 18 (`ge=18`).
2. **Función registrar_usuario:** Cree la función `registrar_usuario(datos: dict) -> str`:
   * Si los datos son válidos, retorne:  
     `"Usuario <username> registrado con éxito (<email>)"`
   * Si no cumplen el esquema y se genera `ValidationError`, capture el error y retorne:  
     `"Registro rechazado: Datos de usuario inválidos"`
