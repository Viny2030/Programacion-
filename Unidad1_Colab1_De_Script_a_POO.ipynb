## 1. ¿Por qué orientación a objetos?

Con pocos datos, funciones y diccionarios alcanzan:

```python
producto = {'nombre': 'Notebook Lenovo', 'precio': 599.0, 'stock': 10}

def aplicar_descuento(producto, porcentaje):
    producto['precio'] *= (1 - porcentaje / 100)
```

A medida que el proyecto crece (más productos, más operaciones, más reglas de validación), este estilo se vuelve difícil de mantener: los datos y el comportamiento que actúa sobre ellos quedan separados, y nada impide que alguien arme un diccionario con las claves mal escritas o incompletas.

La **programación orientada a objetos (POO)** agrupa datos (atributos) y comportamiento (métodos) en una misma unidad: la **clase**.

````python
producto = {'nombre': 'Notebook Lenovo', 'precio': 599.0, 'stock': 10}

def aplicar_descuento(producto, porcentaje):
    producto['precio'] *= (1 - porcentaje / 100)
````

## 2. Clases y objetos

- Una **clase** es la plantilla/definición (`Producto`).
- Un **objeto** (o **instancia**) es algo concreto creado a partir de esa clase (`notebook = Producto(...)`).
- `__init__` es el constructor: se ejecuta al crear una instancia, y ahí se inicializan los atributos de instancia (`self.nombre`, `self.precio`).

Documentación oficial: [Clases (tutorial de Python)](https://docs.python.org/3/tutorial/classes.html)

## 3. Atributos de instancia vs. atributos de clase

```python
class Producto:
    iva = 0.21  # atributo de clase: compartido por todas las instancias

    def __init__(self, nombre, precio):
        self.nombre = nombre    # atributo de instancia: propio de cada objeto
        self.precio = precio

p1 = Producto('Mouse', 15.0)
p2 = Producto('Teclado', 45.0)

print(p1.iva, p2.iva)        # 0.21 0.21 (comparten el mismo atributo de clase)
print(p1.nombre, p2.nombre)  # cada uno con su propio valor
```

Cambiar `Producto.iva` afecta a todas las instancias; cambiar `p1.precio` solo afecta a `p1`.

````python
class Producto:
    iva = 0.21  # atributo de clase: compartido por todas las instancias

    def __init__(self, nombre, precio):
        self.nombre = nombre    # atributo de instancia: propio de cada objeto
        self.precio = precio

p1 = Producto('Mouse', 15.0)
p2 = Producto('Teclado', 45.0)

print(p1.iva, p2.iva)        # 0.21 0.21 (comparten el mismo atributo de clase)
print(p1.nombre, p2.nombre)  # cada uno con su propio valor
````

## 4. Métodos de instancia y `self`

Cada método de una clase recibe `self` como primer parámetro: es una referencia a la instancia concreta sobre la que se llamó al método. `objeto.metodo(x)` es, en el fondo, azúcar sintáctica para `Clase.metodo(objeto, x)`.

```python
class Carrito:
    def __init__(self):
        self.items = []

    def agregar(self, producto):
        self.items.append(producto)

    def total(self):
        return sum(p.precio for p in self.items)
```

## 5. Muchas instancias, cada una con su propio estado

```python
carrito_ana = Carrito()
carrito_bruno = Carrito()

carrito_ana.agregar(Producto('Notebook', 599.0))

print(carrito_ana.cantidad_items())    # 1
print(carrito_bruno.cantidad_items())  # 0, no se ven afectados entre si
```

Esta es la ventaja central de la POO frente a diccionarios sueltos: cada objeto mantiene su propio estado de forma aislada, con un comportamiento (métodos) definido una sola vez en la clase.

````python
class Producto:
    def __init__(self, nombre: str, precio: float):
        self.nombre = nombre
        self.precio = precio


class Carrito:
    def __init__(self):
        # Lista por instancia: asegura que los carritos sean independientes
        self.productos = []

    def agregar(self, producto: Producto):
        self.productos.append(producto)

    def cantidad_items(self) -> int:
        return len(self.productos)


# --- Prueba de ejecución ---
carrito_ana = Carrito()
carrito_bruno = Carrito()

carrito_ana.agregar(Producto('Notebook', 599.0))

print(carrito_ana.cantidad_items())    # 1
print(carrito_bruno.cantidad_items())  # 0, no se ven afectados entre sí
````

## Mini-proyecto: clase `Producto` completa

Diseñá una clase `Producto` con:

1. Atributos de instancia: `nombre`, `precio`, `stock`.
2. Un atributo de clase `iva = 0.21`.
3. Un método `precio_con_iva(self)` que devuelva `precio * (1 + iva)`.
4. Un método `vender(self, cantidad)` que descuente `cantidad` de `stock`, y no lo deje bajar de cero (si `cantidad > stock`, vender solo lo disponible).
5. Creá 3 instancias distintas y probá cada método.

**Entregable:** la clase completa + las 3 instancias con al menos una llamada a cada método.

---

**Seguís en:** *Colab 2 — Principios de POO y métodos especiales*

````python
class Producto:
    iva = 0.21  # Atributo de clase

    def __init__(self, nombre: str, precio: float, stock: int):
        # Atributos de instancia
        self.nombre = nombre
        self.precio = precio
        self.stock = stock

    def precio_con_iva(self) -> float:
        return self.precio * (1 + self.iva)

    def vender(self, cantidad: int) -> int:
        if cantidad <= self.stock:
            unidades_vendidas = cantidad
            self.stock -= cantidad
        else:
            unidades_vendidas = self.stock
            self.stock = 0
        return unidades_vendidas


# --- 3 Instancias distintas ---
prod1 = Producto("Mouse inalámbrico", 25.0, 10)
prod2 = Producto("Teclado mecánico", 80.0, 3)
prod3 = Producto("Monitor 24 pulgadas", 150.0, 0)

# --- Pruebas de los métodos ---

# Producto 1: Venta normal (hay stock suficiente)
print(f"Producto: {prod1.nombre}")
print(f"Precio con IVA: ${prod1.precio_con_iva():.2f}")
vendidos = prod1.vender(4)
print(f"Unidades vendidas: {vendidos} | Stock restante: {prod1.stock}\n")

# Producto 2: Se pide más de lo disponible (vende solo lo que hay y no baja de cero)
print(f"Producto: {prod2.nombre}")
print(f"Precio con IVA: ${prod2.precio_con_iva():.2f}")
vendidos = prod2.vender(5)
print(f"Unidades vendidas: {vendidos} | Stock restante: {prod2.stock}\n")

# Producto 3: Se intenta vender sin stock inicial
print(f"Producto: {prod3.nombre}")
print(f"Precio con IVA: ${prod3.precio_con_iva():.2f}")
vendidos = prod3.vender(2)
print(f"Unidades vendidas: {vendidos} | Stock restante: {prod3.stock}")
````
